```python
# !pip install python-docx -q

import os
import glob
from docx import Document

input_dir = '/kaggle/input/datasets/zahidazam/cpt-dataset/cpt-data-text/'
docx_files = glob.glob(os.path.join(input_dir, '*.docx'))
output_file = '/kaggle/working/khowar_docx.txt'

total_chars = 0
with open(output_file, 'w', encoding='utf-8') as out_f:
    for file_path in docx_files:
        print(f"Processing: {os.path.basename(file_path)}")
        doc = Document(file_path)
        for para in doc.paragraphs:
            if para.text.strip():
                out_f.write(para.text + '\n')
                total_chars += len(para.text)

print(f"Extracted text from {len(docx_files)} files.")
print(f"Total characters from docx: {total_chars}")
```

```python
import glob
import os

input_dir = '/kaggle/input/datasets/zahidazam/cpt-dataset/cpt-data-text/'
txt_files = glob.glob(os.path.join(input_dir, '*.txt'))
output_file = '/kaggle/working/khowar_combined.txt'

total_chars = 0
with open(output_file, 'w', encoding='utf-8') as out_f:
    for file_path in txt_files:
        with open(file_path, 'r', encoding='utf-8') as in_f:
            content = in_f.read()
            out_f.write(content + '\n\n')
            total_chars += len(content)

print(f"Combined {len(txt_files)} files.")
print(f"Total characters: {total_chars}")
```

```python
txt_file = '/kaggle/working/khowar_combined.txt'
docx_file = '/kaggle/working/khowar_docx.txt'
final_file = '/kaggle/working/khowar_final.txt'

with open(txt_file, 'r', encoding='utf-8') as t, open(docx_file, 'r', encoding='utf-8') as d:
    txt_chars = len(t.read())
    d.seek(0)
    docx_chars = len(d.read())

# Read and merge
with open(txt_file, 'r', encoding='utf-8') as t, open(docx_file, 'r', encoding='utf-8') as d, open(final_file, 'w', encoding='utf-8') as f:
    f.write(t.read())
    f.write('\n\n')
    f.write(d.read())

print(f"Total characters: {txt_chars + docx_chars}")
print("Merged into khowar_final.txt!")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import torch

model_id = "Qwen/Qwen2.5-0.5B"

# 4-bit quantization config for QLoRA
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token

# Load model
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto",
)

print("Model loaded successfully!")
print(f"Model device: {next(model.parameters()).device}")
```

```python
from datasets import load_dataset

# 1. Text file ko load karein
dataset = load_dataset('text', data_files='/kaggle/working/khowar_final.txt', split='train')
block_size = 1024

# 2. Tokenize karne ka function
def tokenize_function(examples):
    return tokenizer(examples["text"])

tokenized_datasets = dataset.map(tokenize_function, batched=True, num_proc=2, remove_columns=["text"])

# 3. Text ko 1024 tokens ke blocks mein divide karna
def group_texts(examples):
    concatenated = {k: sum(examples[k], []) for k in examples.keys()}
    total_length = (len(concatenated['input_ids']) // block_size) * block_size
    
    result = {
        k: [t[i : i + block_size] for i in range(0, total_length, block_size)]
        for k, t in concatenated.items()
    }
    result["labels"] = result["input_ids"].copy()
    return result

# 4. Apply grouping
lm_datasets = tokenized_datasets.map(group_texts, batched=True, num_proc=2)

print(f"Total training samples (chunks): {len(lm_datasets)}")
```

```python
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
from transformers import DataCollatorForLanguageModeling, TrainingArguments, Trainer

# 1. LoRA Config for Qwen 2.5
lora_config = LoraConfig(
    r=16, lora_alpha=32,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_dropout=0.05, task_type="CAUSAL_LM"
)

model = prepare_model_for_kbit_training(model)
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()

# 2. Trainer setup
training_args = TrainingArguments(
    output_dir="/kaggle/working/qwen-khowar-cpt",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    num_train_epochs=3,
    logging_steps=20,
    save_strategy="epoch",
    bf16=True,
    optim="paged_adamw_8bit",
)

trainer = Trainer(
    model=model, args=training_args,
    train_dataset=lm_datasets,
    data_collator=DataCollatorForLanguageModeling(tokenizer, mlm=False),
)
```

```python
model_id = "Qwen/Qwen2.5-0.5B"

# Force load model on GPU 0 only
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map={"": 0},
)

model = prepare_model_for_kbit_training(model)
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()

trainer = Trainer(
    model=model, args=training_args,
    train_dataset=lm_datasets,
    data_collator=DataCollatorForLanguageModeling(tokenizer, mlm=False),
)
```
